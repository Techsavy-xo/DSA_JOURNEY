# 975. Pattern 3 – Number Triangle

## Problem Statement

Given an integer `n`, print a pattern of numbers where each row starts from `1` and continues up to the row number.

For example, if `n = 5`, the pattern should be:

```text
1
12
123
1234
12345
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
1
12
123
1234
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
1
12
```

## Explanation

- `for i in range(n):` controls the number of rows.
- `for j in range(1, i + 2):` prints numbers starting from `1` up to the current row number.
- `print(j, end="")` prints numbers on the same line without spaces.
- `print()` moves to the next line after each row.

## Dry Run for n = 4

| `i` | `j` values | Output |
|---:|---|---|
| 0 | 1 | `1` |
| 1 | 1, 2 | `12` |
| 2 | 1, 2, 3 | `123` |
| 3 | 1, 2, 3, 4 | `1234` |

## Key Concept

**Remember:** The outer loop controls the rows, while the inner loop prints numbers from `1` to the current row number.

We use `i + 2` because `range()` excludes its ending value, and `i` starts from `0`.

## Python Solution

```python
class Solution:
    def pattern3(self, n):
        for i in range(n):
            for j in range(1, i + 2):
                print(j, end="")
            print()
```
