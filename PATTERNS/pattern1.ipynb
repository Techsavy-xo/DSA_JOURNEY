## Problem Statement
Given an integer `n`, print a square pattern of stars (`*`) with `n` rows and `n` columns.

### Example 1
**Input:**
```python
n = 4
```

**Output:**
```text
* * * *
* * * *
* * * *
* * * *
```

### Example 2
**Input:**
```python
n = 2
```

**Output:**
```text
* *
* *
```

### Example 3
**Input:**
```python
n = 5
```

**Output:**
```text
* * * * *
* * * * *
* * * * *
* * * * *
* * * * *
```

### Constraints
- `1 <= n <= 100`

### Approach
1. Use an outer loop to iterate through `n` rows.
2. Use an inner loop to print `n` stars in each row.
3. Move to the next line after printing each row.
