# Pattern 9 – Diamond Star Pattern

## Problem Statement

Given an integer `n`, print a diamond pattern using stars (`*`).

The upper half forms a pyramid, with stars increasing by two in each row. The lower half forms an inverted pyramid, with stars decreasing by two in each row.

For example, if `n = 5`, the pattern should be:

```text
    *
   ***
  *****
 *******
*********
*********
 *******
  *****
   ***
    *
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
   *
  ***
 *****
*******
*******
 *****
  ***
   *
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
 *
***
***
 *
```

## Explanation

### Upper half — Pyramid

- `for i in range(1, n + 1):` runs from the first row to the last row of the pyramid.
- `spaces = n - i` decreases the leading spaces by one each row.
- `stars = 2 * i - 1` increases the stars by two each row.
- `print(" " * spaces + "*" * stars)` prints the spaces followed by the stars.

### Lower half — Inverted pyramid

- `for j in range(n, 0, -1):` runs backward from `n` to `1`.
- `space = n - j` increases the leading spaces by one each row.
- `star = 2 * j - 1` decreases the stars by two each row.
- `print(" " * space + "*" * star)` prints each row of the inverted pyramid.

**Important:** Your current code prints the widest row twice because both loops include the middle row. This matches the pattern shown in the question.

## Dry Run for n = 3

| Loop | Row | Spaces | Stars | Output |
|---|---:|---:|---:|---|
| Upper | 1 | 2 | 1 | `  *` |
| Upper | 2 | 1 | 3 | ` ***` |
| Upper | 3 | 0 | 5 | `*****` |
| Lower | 3 | 0 | 5 | `*****` |
| Lower | 2 | 1 | 3 | ` ***` |
| Lower | 1 | 2 | 1 | `  *` |

## Key Concepts

**Memory trick:**

- Upper half: spaces decrease, stars increase.
- Lower half: spaces increase, stars decrease.
- Spaces formula: `n - i` in the upper half and `n - j` in the lower half.
- Stars formula: `2 * row - 1`.

## Python Solution

```python
class Solution:
    def pattern9(self, n):
        for i in range(1, n + 1):
            spaces = n - i
            stars = 2 * i - 1
            print(" " * spaces + "*" * stars)

        for j in range(n, 0, -1):
            space = n - j
            star = 2 * j - 1
            print(" " * space + "*" * star)
```
