# Pattern 8 – Inverted Pyramid Star Pattern

## Problem Statement

Given an integer `n`, print an inverted pyramid pattern using stars (`*`). The first row contains `2 * n - 1` stars, and each subsequent row contains two fewer stars than the previous row. Spaces increase by one on each row to keep the pattern centered.

For example, if `n = 5`, the pattern should be:

```text
*********
 *******
  *****
   ***
    *
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
*******
 *****
  ***
   *
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
***
 *
```

## Explanation

- `for i in range(1, n + 1):` controls the rows, from `1` to `n`.
- `space = i - 1` increases the leading spaces by one on every row.
- `star = 2 * n - 1 - 2 * (i - 1)` calculates the number of stars, decreasing by two on each row.
- `" " * space` repeats the spaces.
- `"*" * star` repeats the stars.
- `print()` displays the spaces and stars together on one line.

## Dry Run for n = 4

| `i` | `space = i - 1` | `star = 2*n - 1 - 2*(i - 1)` | Output |
|---:|---:|---:|---|
| 1 | 0 | 7 | `*******` |
| 2 | 1 | 5 | ` *****` |
| 3 | 2 | 3 | `  ***` |
| 4 | 3 | 1 | `   *` |

## Key Concepts

**1. Spaces increase:** Each new row adds one leading space.

**2. Stars decrease:** Each new row removes two stars.

**Memory trick:**
- Spaces = `i - 1`
- Stars = `2 * n - 1 - 2 * (i - 1)`

## Python Solution

```python
class Solution:
    def pattern8(self, n):
        for i in range(1, n + 1):
            space = i - 1
            star = 2 * n - 1 - 2 * (i - 1)
            print(" " * space + "*" * star)
```
