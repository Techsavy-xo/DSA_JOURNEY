# Pattern 11 – Binary Number Triangle

## 1. Problem Statement

Given an integer `n`, print a triangle of `0`s and `1`s.

The pattern alternates between `0` and `1` in every row, and the starting number changes in each row.

### Example 1

**Input:** `n = 4`

**Output:**
```text
1
0 1
1 0 1
0 1 0 1
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
1
0 1
```

---

## 2. Understanding the Logic

Your code uses **two loops** and the modulo operator `%`.

### Step 1: The outer loop

```python
for i in range(1, n + 1):
```

This controls the rows.

- `i = 1` → first row
- `i = 2` → second row
- `i = 3` → third row

The outer loop runs from `1` to `n`.

### Step 2: The inner loop

```python
for j in range(i):
```

This controls how many numbers appear in each row.

Remember, `range(i)` starts at `0`.

| `i` | Values of `j` | Numbers printed |
|---:|---|---:|
| 1 | 0 | 1 number |
| 2 | 0, 1 | 2 numbers |
| 3 | 0, 1, 2 | 3 numbers |
| 4 | 0, 1, 2, 3 | 4 numbers |

So, the row number tells us how many numbers to print.

### Step 3: The modulo formula

```python
oprt = (i + j) % 2
```

This is the most important line!

The `%` operator gives the **remainder after division**.

- An even number `% 2` gives `0`.
- An odd number `% 2` gives `1`.

Therefore, `(i + j) % 2` produces alternating `0`s and `1`s.

Let's check the first few calculations:

| `i` | `j` | `i + j` | `(i + j) % 2` |
|---:|---:|---:|---:|
| 1 | 0 | 1 | 1 |
| 2 | 0 | 2 | 0 |
| 2 | 1 | 3 | 1 |
| 3 | 0 | 3 | 1 |
| 3 | 1 | 4 | 0 |
| 3 | 2 | 5 | 1 |

Notice the pattern:

- Row 1 starts with `1`.
- Row 2 starts with `0`.
- Row 3 starts with `1`.
- Row 4 starts with `0`.

The formula automatically creates this behavior.

### Step 4: Printing on the same line

```python
print(oprt, end=" ")
```

Normally, `print()` moves to a new line after printing.

Using `end=" "` adds a space instead of moving to the next line.

```python
print()
```

After the inner loop finishes, this moves to the next row.

---

## 3. Dry Run for `n = 3`

**Row 1:** `i = 1`

- `j = 0` → `(1 + 0) % 2 = 1`

Output:
```text
1
```

**Row 2:** `i = 2`

- `j = 0` → `(2 + 0) % 2 = 0`
- `j = 1` → `(2 + 1) % 2 = 1`

Output:
```text
0 1
```

**Row 3:** `i = 3`

- `j = 0` → `(3 + 0) % 2 = 1`
- `j = 1` → `(3 + 1) % 2 = 0`
- `j = 2` → `(3 + 2) % 2 = 1`

Output:
```text
1 0 1
```

### Final pattern

```text
1
0 1
1 0 1
```

---

## 4. Memory Trick

Remember **Rows → Positions → Modulo**.

1. Outer loop `i` controls the row.
2. Inner loop `j` controls the position.
3. `(i + j) % 2` decides whether to print `0` or `1`.

**Easy rule:** Even sum gives `0`; odd sum gives `1`.

---

## 5. Python Solution

```python
class Solution:
    def pattern11(self, n):
        for i in range(1, n + 1):
            for j in range(i):
                oprt = (i + j) % 2
                print(oprt, end=" ")
            print()
```
