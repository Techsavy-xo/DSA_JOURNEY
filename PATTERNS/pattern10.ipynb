# Pattern 10 – Increasing and Decreasing Stars

## 1. Problem Statement

Given an integer `n`, print a pattern where:

- The first half prints stars in increasing order, from `1` to `n`.
- The second half prints stars in decreasing order, from `n-1` to `1`.

### Example 1

**Input:** `n = 4`

**Output:**
```text
*
**
***
****
***
**
*
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
*
**
*
```

---

## 2. Understanding the Logic

We need **two loops** because the number of stars first increases and then decreases.

### First loop — Increasing stars

```python
for i in range(1, n + 1):
    print("*" * i)
```

- `range(1, n + 1)` starts at `1` and ends at `n`.
- `"*" * i` prints the star `i` times.
- Each iteration prints one row.

For `n = 4`:

| `i` | Stars printed |
|---:|---|
| 1 | `*` |
| 2 | `**` |
| 3 | `***` |
| 4 | `****` |

### Second loop — Decreasing stars

```python
for j in range(n - 1, 0, -1):
    print("*" * j)
```

- `n - 1` starts from one less than `n`.
- `0` is the stopping point and is not included.
- `-1` means decrease by one each time.

For `n = 4`:

| `j` | Stars printed |
|---:|---|
| 3 | `***` |
| 2 | `**` |
| 1 | `*` |

**Why start from `n - 1` instead of `n`?**

Because the first loop has already printed the widest row, `****`. Starting from `n` again would print that row twice.

---

## 3. Dry Run for `n = 3`

**First loop:**

```text
*
**
***
```

**Second loop:**

```text
**
*
```

**Final output:**

```text
*
**
***
**
*
```

---

## 4. Memory Trick

Remember: **Up → Down**

- First loop: `1` to `n` — stars increase.
- Second loop: `n-1` to `1` — stars decrease.
- `* i` means print `i` stars.
- `-1` in `range()` means count backward.

## 5. Python Solution

```python
class Solution:
    def pattern10(self, n):
        for i in range(1, n + 1):
            print("*" * i)

        for j in range(n - 1, 0, -1):
            print("*" * j)
```
