# Pattern 14 – Increasing Alphabet Triangle

## 1. Problem Statement

Given an integer `n`, print a triangle of alphabets starting from `A`.

- First row prints `A`.
- Second row prints `AB`.
- Third row prints `ABC`.
- Each new row adds one more alphabet.

### Example 1

**Input:** `n = 4`

**Output:**
```text
A
AB
ABC
ABCD
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
A
AB
```

---

## 2. Understanding the Logic

### Step 1: Outer loop – Controls rows

```python
for i in range(1, n + 1):
```

The outer loop decides how many rows to print.

For `n = 5`, `i` takes the values:

`1, 2, 3, 4, 5`

### Step 2: Inner loop – Controls alphabets

```python
for j in range(i):
```

The inner loop prints as many alphabets as the current row number.

| Row (`i`) | Values of `j` | Alphabets |
|---:|---|---|
| 1 | 0 | A |
| 2 | 0, 1 | AB |
| 3 | 0, 1, 2 | ABC |
| 4 | 0, 1, 2, 3 | ABCD |
| 5 | 0, 1, 2, 3, 4 | ABCDE |

### Step 3: Understanding `chr()` and ASCII

```python
print(chr(65 + j), end="")
```

This is the most important line!

Python uses ASCII values to represent English characters.

| Character | ASCII value |
|---|---:|
| A | 65 |
| B | 66 |
| C | 67 |
| D | 68 |
| E | 69 |
| F | 70 |

**What does `chr()` do?**

`chr()` converts an integer into its corresponding character.

For example:

```python
chr(65)  # A
chr(66)  # B
chr(67)  # C
```

Now let's understand your formula:

`chr(65 + j)`

| `j` | Calculation | Output |
|---:|---|---|
| 0 | 65 + 0 | A |
| 1 | 65 + 1 | B |
| 2 | 65 + 2 | C |
| 3 | 65 + 3 | D |
| 4 | 65 + 4 | E |

Since `j` starts at zero, adding it to 65 gives us the correct alphabet.

### Step 4: Understanding `end=""`

```python
print(chr(65 + j), end="")
```

Normally, `print()` moves to a new line after printing.

`end=""` prevents that, allowing all alphabets in the same row to appear together.

```python
print()
```

This moves to the next row after the inner loop finishes.

## 3. Dry Run for n = 4

**Row 1:** `i = 1`

- `j = 0` → `chr(65 + 0)` → A

Output: `A`

**Row 2:** `i = 2`

- `j = 0` → A
- `j = 1` → B

Output: `AB`

**Row 3:** `i = 3`

- `j = 0` → A
- `j = 1` → B
- `j = 2` → C

Output: `ABC`

**Row 4:** `i = 4`

- `j = 0` → A
- `j = 1` → B
- `j = 2` → C
- `j = 3` → D

Output: `ABCD`

### Final Pattern

```text
A
AB
ABC
ABCD
```

---

## 4. Memory Trick

Remember: **Rows → Positions → ASCII → Character**

1. Outer loop `i` controls the rows.
2. Inner loop `j` controls the number of alphabets.
3. `65` represents uppercase A.
4. `chr(65 + j)` converts the ASCII value into an alphabet.
5. `end=""` keeps the alphabets on the same line.

**Important:** `ord('A')` gives `65`, while `chr(65)` gives `'A'`.

---

## 5. Python Solution

```python
class Solution:
    def pattern14(self, n):
        for i in range(1, n + 1):
            for j in range(i):
                print(chr(65 + j), end="")
            print()
```
