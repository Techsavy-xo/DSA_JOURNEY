# 987. Pattern 6 – Decreasing Number Triangle

## Problem Statement

Given an integer `n`, print a pattern where the first row contains numbers from `1` to `n`. Each subsequent row contains one fewer number, until the last row contains only `1`.

For example, if `n = 5`, the pattern should be:

```text
12345
1234
123
12
1
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
1234
123
12
1
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
12
1
```

## Explanation

- `for i in range(n, 0, -1):` controls the rows, starting from `n` and decreasing to `1`.
- `for j in range(1, i + 1):` prints numbers from `1` up to `i`.
- `print(j, end="")` prints each number on the same line.
- `print()` moves to the next line after each row.

## Dry Run for n = 4

| `i` | `j` values | Output |
|---:|---|---|
| 4 | 1, 2, 3, 4 | `1234` |
| 3 | 1, 2, 3 | `123` |
| 2 | 1, 2 | `12` |
| 1 | 1 | `1` |

## Key Concept

**Remember:** The outer loop decreases the row length, while the inner loop prints numbers starting from `1` up to the current value of `i`.

For example, when `i = 3`, `range(1, i + 1)` becomes `range(1, 4)`, which produces `1, 2, 3`.

## Python Solution

```python
class Solution:
    def pattern6(self, n):
        for i in range(n, 0, -1):
            for j in range(1, i + 1):
                print(j, end="")
            print()
```
