# 986. Pattern 5 – Decreasing Star Triangle

## Problem Statement

Given an integer `n`, print a pattern of stars (`*`) where the first row contains `n` stars, and each subsequent row contains one fewer star than the previous row.

For example, if `n = 5`, the pattern should be:

```text
*****
****
***
**
*
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
****
***
**
*
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
**
*
```

## Explanation

- `for i in range(n, 0, -1):` controls the rows, starting from `n` and decreasing to `1`.
- `for j in range(0, i):` prints `i` stars in each row.
- `print("*", end="")` prints each star on the same line.
- `print()` moves to the next line after each row.

## Dry Run for n = 4

| `i` | `j` values | Number of stars | Output |
|---:|---|---:|---|
| 4 | 0, 1, 2, 3 | 4 | `****` |
| 3 | 0, 1, 2 | 3 | `***` |
| 2 | 0, 1 | 2 | `**` |
| 1 | 0 | 1 | `*` |

## Key Concept

**Remember:** The outer loop decreases from `n` to `1`, while the inner loop prints exactly `i` stars.

`range(n, 0, -1)` means:
- Start at `n`.
- Stop before `0`.
- Decrease by `1` each time.

## Python Solution

```python
class Solution:
    def pattern5(self, n):
        for i in range(n, 0, -1):
            for j in range(0, i):
                print("*", end="")
            print()
```
