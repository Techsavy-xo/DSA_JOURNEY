# Pattern 4 – Number Triangle

## Problem Statement

Given an integer `n`, print a pattern where each row contains the same number repeated. The number corresponds to the current row, and the number of repetitions equals the row number.

For example, if `n = 5`, the pattern should be:

```text
1
22
333
4444
55555
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
1
22
333
4444
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
1
22
```

## Explanation

- `for i in range(n):` controls the number of rows.
- `for j in range(1, i + 2):` repeats the inner loop according to the current row.
- `print(i + 1, end="")` prints the same number repeatedly on the current line.
- `print()` moves to the next line after each row.

## Dry Run for n = 4

| `i` | Number printed | Repetitions | Output |
|---:|---:|---:|---|
| 0 | 1 | 1 | `1` |
| 1 | 2 | 2 | `22` |
| 2 | 3 | 3 | `333` |
| 3 | 4 | 4 | `4444` |

## Key Concept

**Remember:** The outer loop controls the rows, the inner loop controls the repetitions, and `i + 1` gives the number to print because `i` starts at `0`.

## Python Solution

```python
class Solution:
    def pattern4(self, n):
        for i in range(n):
            for j in range(1, i + 2):
                print(i + 1, end="")
            print()
```
