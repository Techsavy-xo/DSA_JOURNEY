# Pattern 7 – Pyramid Star Pattern

## Problem Statement

Given an integer `n`, print a pyramid pattern using stars (`*`). Each row contains an increasing number of stars, with spaces before the stars to keep the pyramid centered.

The number of stars in each row follows the sequence `1, 3, 5, 7, 9, ...`.

For example, if `n = 5`, the pattern should be:

```text
    *
   ***
  *****
 *******
*********
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
   *
  ***
 *****
*******
```

## Example 2

**Input:**
```text
n = 2
```

**Output:**
```text
 *
***
```

## Explanation

- `for i in range(1, n + 1):` controls the rows, starting from `1` and ending at `n`.
- `spaces = n - i` calculates the number of spaces before the stars.
- `stars = 2 * i - 1` calculates the number of stars in each row.
- `" " * spaces` repeats the space character.
- `"*" * stars` repeats the star character.
- `print(" " * spaces + "*" * stars)` prints the spaces and stars together on the same line.

## Dry Run for n = 4

| `i` | `spaces = n - i` | `stars = 2 * i - 1` | Output |
|---:|---:|---:|---|
| 1 | 3 | 1 | `   *` |
| 2 | 2 | 3 | `  ***` |
| 3 | 1 | 5 | ` *****` |
| 4 | 0 | 7 | `*******` |

## Key Concepts

**1. Spaces decrease:** As the row number increases, the leading spaces decrease by one.

**2. Stars increase by two:** Each new row adds two stars compared with the previous row.

**Memory trick:**
- Spaces = `n - i`
- Stars = `2 * i - 1`

## Python Solution

```python
class Solution:
    def pattern7(self, n):
        for i in range(1, n + 1):
            spaces = n - i
            stars = 2 * i - 1
            print(" " * spaces + "*" * stars)
```
