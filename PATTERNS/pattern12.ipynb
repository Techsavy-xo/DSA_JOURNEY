# Pattern 12 – Number Triangle with Spaces

## 1. Problem Statement

Given an integer `n`, print a pattern where:

- The left side prints numbers in increasing order.
- The middle contains spaces that decrease with every row.
- The right side prints numbers in decreasing order.

### Example 1

**Input:** `n = 4`

**Output:**
```text
1      1
12    21
123  321
12344321
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
1  1
1221
```

---

## 2. Understanding the Logic

We use one outer loop and two inner loops, along with a space calculation.

### Step 1: Outer loop – Rows

```python
for i in range(1, n + 1):
```

This controls the number of rows.

For `n = 4`, `i` takes these values:

`1, 2, 3, 4`

### Step 2: Left side – Increasing numbers

```python
for j in range(1, i + 1):
    print(j, end="")
```

This prints numbers from `1` to `i`.

| Row (`i`) | Numbers |
|---:|---|
| 1 | 1 |
| 2 | 12 |
| 3 | 123 |
| 4 | 1234 |

### Step 3: Middle – Spaces

```python
space = 2 * (n - i)
print(" " * space, end="")
```

This is the important formula!

Why `2 * (n - i)`?

- `n - i` calculates how many rows are left after the current row.
- Multiplying by `2` gives the middle spacing needed for the pattern.

For `n = 4`:

| Row (`i`) | Calculation | Spaces |
|---:|---|---:|
| 1 | 2 × (4 - 1) | 6 |
| 2 | 2 × (4 - 2) | 4 |
| 3 | 2 × (4 - 3) | 2 |
| 4 | 2 × (4 - 4) | 0 |

Notice that the spaces decrease by 2 in every row.

### Step 4: Right side – Decreasing numbers

```python
for j in range(i, 0, -1):
    print(j, end="")
```

This prints numbers backwards, starting from `i` and ending at `1`.

| Row (`i`) | Numbers |
|---:|---|
| 1 | 1 |
| 2 | 21 |
| 3 | 321 |
| 4 | 4321 |

Remember:
- Start: `i`
- Stop: `0` (not included)
- Step: `-1` (count backwards)

### Step 5: Move to the next row

```python
print()
```

This moves the cursor to the next line after completing both number sequences.

---

## 4. Memory Trick

Remember: **Left → Space → Right**

- Left: `range(1, i + 1)` → increasing.
- Space: `2 * (n - i)` → decreasing gaps.
- Right: `range(i, 0, -1)` → decreasing.

The outer loop controls the rows, while the inner loops control the numbers.

---

## 5. Python Solution

```python
class Solution:
    def pattern12(self, n):
        for i in range(1, n + 1):

            # Left side: increasing numbers
            for j in range(1, i + 1):
                print(j, end="")

            # Middle: decreasing spaces
            space = 2 * (n - i)
            print(" " * space, end="")

            # Right side: decreasing numbers
            for j in range(i, 0, -1):
                print(j, end="")

            print()
```
