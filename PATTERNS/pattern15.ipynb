# Pattern 15 – Decreasing Alphabet Triangle

## 1. Problem Statement

Given an integer `n`, print a triangle of alphabets in decreasing order of row length.

- First row prints `n` alphabets.
- Every next row prints one fewer alphabet.
- The alphabets always start from `A`.

### Example 1

**Input:** `n = 4`

**Output:**
```text
ABCD
ABC
AB
A
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
AB
A
```

---

## 2. Understanding the Logic

We use two loops:

- Outer loop controls the rows and decreases.
- Inner loop controls how many alphabets to print.

### Step 1: Outer loop – Decreasing rows

```python
for i in range(n, 0, -1):
```

Let's understand the `range()`:

`range(start, stop, step)`

- Start: `n`
- Stop: `0` (not included)
- Step: `-1` (decrease by one)

For `n = 5`, the values of `i` are:

`5, 4, 3, 2, 1`

This means the first row has 5 alphabets, the second has 4, and so on.

### Step 2: Inner loop – Printing alphabets

```python
for j in range(i):
    print(chr(65 + j), end="")
```

The inner loop starts at `j = 0` every time.

Remember the ASCII values:

| `j` | Calculation | Character |
|---:|---|---|
| 0 | 65 + 0 | A |
| 1 | 65 + 1 | B |
| 2 | 65 + 2 | C |
| 3 | 65 + 3 | D |
| 4 | 65 + 4 | E |

Notice that the alphabet always starts from `A`, but the number of alphabets decreases.

### Step 3: Move to the next row

```python
print()
```

After the inner loop finishes, this moves to the next line.

---

## 3. Dry Run for n = 4

| Row | `i` | `j` values | Output |
|---:|---:|---|---|
| 1 | 4 | 0, 1, 2, 3 | ABCD |
| 2 | 3 | 0, 1, 2 | ABC |
| 3 | 2 | 0, 1 | AB |
| 4 | 1 | 0 | A |

### Final Output

```text
ABCD
ABC
AB
A
```

---

## 4. Important Concept – Difference Between Pattern 14 and 15

| Feature | Pattern 14 | Pattern 15 |
|---|---|---|
| Outer loop | `range(1, n + 1)` | `range(n, 0, -1)` |
| Row length | Increases | Decreases |
| First row (`n=5`) | A | ABCDE |
| Last row (`n=5`) | ABCDE | A |

**Key observation:** We only changed the outer loop. The inner loop and ASCII formula remain the same!

---

## 5. Memory Trick

Remember: **Reverse the outer loop, keep the inner loop.**

- `range(n, 0, -1)` → rows decrease.
- `range(i)` → prints `i` alphabets.
- `chr(65 + j)` → starts from A every row.
- `end=""` → prints alphabets together.

---

## 6. Python Solution

```python
class Solution:
    def pattern15(self, n):
        for i in range(n, 0, -1):
            for j in range(i):
                print(chr(65 + j), end="")
            print()
```
