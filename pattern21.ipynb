# Pattern 21 – Hollow Square Pattern

## Problem Statement

Given an integer `n`, print a hollow square pattern of stars (`*`). The first and last rows should contain only stars. The middle rows should contain stars at the beginning and end, with spaces in between.

## Python Solution

```python
class Solution:
    def pattern21(self, n):
        for i in range(n):
            for j in range(n):
                if i == 0 or i == n - 1 or j == 0 or j == n - 1:
                    print("*", end="")
                else:
                    print(" ", end="")
            print()
```

## Example 1

**Input:**
```text
n = 4
```

**Output:**
```text
****
*  *
*  *
****
```

## Example 2

**Input:**
```text
n = 5
```

**Output:**
```text
*****
*   *
*   *
*   *
*****
```

## Explanation

- `range(n)` runs the outer loop `n` times, once for each row.
- `range(n)` runs the inner loop `n` times, once for each column.
- `i == 0` checks whether it is the first row.
- `i == n - 1` checks whether it is the last row.
- `j == 0` checks whether it is the first column.
- `j == n - 1` checks whether it is the last column.
- If any condition is true, print `*`; otherwise, print a space.
- `print()` moves to the next row.

## Dry Run for n = 4

| `i` | `j` values | Stars printed |
|---|---|---|
| 0 | 0, 1, 2, 3 | `****` |
| 1 | 0, 1, 2, 3 | `*  *` |
| 2 | 0, 1, 2, 3 | `*  *` |
| 3 | 0, 1, 2, 3 | `****` |

**Remember:** The outer loop controls rows, the inner loop controls columns, and the `if` condition checks whether a position is on the boundary of the square.
