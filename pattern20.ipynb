# Pattern 20 – Butterfly Pattern (Increasing and Decreasing)

## 1. Problem Statement

Given an integer `n`, print a butterfly-like pattern using stars.

- Upper half: Stars increase and middle spaces decrease.
- Lower half: Stars decrease and middle spaces increase.
- Both sides of every row contain the same number of stars.

### Example 1

**Input:** `n = 4`

**Output:**
```text
*      *
**    **
***  ***
********
***  ***
**    **
*      *
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
*  *
****
*  *
```

---

## 2. Understanding the Logic

We use two loops:

1. First loop creates the increasing half.
2. Second loop creates the decreasing half.

Every row follows this structure:

`Left stars + Middle spaces + Right stars`

### Step 1: Upper half – Increasing stars

```python
for i in range(1, n + 1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
```

Let's divide the formula into three parts.

**Left stars:**
```python
"*" * i
```
The number of stars increases with every row.

**Middle spaces:**
```python
" " * (2 * (n - i))
```
The spaces decrease by 2 with every row.

**Right stars:**
```python
"*" * i
```
The right side prints the same number of stars as the left.

For `n = 4`:

| Row (`i`) | Left stars | Middle spaces | Right stars |
|---:|---:|---:|---:|
| 1 | 1 | 6 | 1 |
| 2 | 2 | 4 | 2 |
| 3 | 3 | 2 | 3 |
| 4 | 4 | 0 | 4 |

### Step 2: Lower half – Decreasing stars

```python
for i in range(n - 1, 0, -1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
```

This loop works in reverse.

Let's understand the range:

`range(n - 1, 0, -1)`

- Start: `n - 1`
- Stop: `0` (not included)
- Step: `-1` (decrease by one)

For `n = 4`, the values of `i` are:

`3, 2, 1`

**Why start from `n - 1` instead of `n`?**

Because the first loop has already printed the widest row (`********`).

Starting from `n` again would print that row twice.

For the lower half:

| Row (`i`) | Left stars | Middle spaces | Right stars |
|---:|---:|---:|---:|
| 3 | 3 | 2 | 3 |
| 2 | 2 | 4 | 2 |
| 1 | 1 | 6 | 1 |

---

## 3. Dry Run for n = 4

### First loop

| `i` | Calculation | Output |
|---:|---|---|
| 1 | 1 star + 6 spaces + 1 star | `*      *` |
| 2 | 2 stars + 4 spaces + 2 stars | `**    **` |
| 3 | 3 stars + 2 spaces + 3 stars | `***  ***` |
| 4 | 4 stars + 0 spaces + 4 stars | `********` |

### Second loop

| `i` | Calculation | Output |
|---:|---|---|
| 3 | 3 stars + 2 spaces + 3 stars | `***  ***` |
| 2 | 2 stars + 4 spaces + 2 stars | `**    **` |
| 1 | 1 star + 6 spaces + 1 star | `*      *` |

### Final Output

```text
*      *
**    **
***  ***
********
***  ***
**    **
*      *
```

---

## 4. Important Concept – Pattern 19 vs Pattern 20

| Feature | Pattern 19 | Pattern 20 |
|---|---|---|
| First half stars | Decreasing | Increasing |
| First half spaces | Increasing | Decreasing |
| Second half stars | Increasing | Decreasing |
| Second half spaces | Decreasing | Increasing |
| Widest row | Appears twice | Appears once |

**Key observation:** Pattern 20 starts with the smallest row and expands towards the middle. Pattern 19 starts with the widest row and contracts.

---

## 5. Memory Trick

Remember: **Expand → Contract**

- First loop: `range(1, n + 1)` → stars increase.
- Second loop: `range(n - 1, 0, -1)` → stars decrease.
- Spaces: `2 * (n - i)`.
- Both sides use `"*" * i`.

The same formula works for both loops. Only the direction of `i` changes!

---

## 6. Python Solution

```python
class Solution:
    def pattern(self, n):

        # Upper half
        for i in range(1, n + 1):
            print("*" * i + " " * (2 * (n - i)) + "*" * i)

        # Lower half
        for i in range(n - 1, 0, -1):
            print("*" * i + " " * (2 * (n - i)) + "*" * i)
```
