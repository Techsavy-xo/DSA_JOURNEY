# Pattern 17 – Alphabet Pyramid

## 1. Problem Statement

Given an integer `n`, print a pyramid of alphabets.

- The pyramid is centered using spaces.
- The left side prints alphabets in increasing order.
- The right side prints alphabets in decreasing order.
- The middle alphabet appears only once.

### Example 1

**Input:** `n = 4`

**Output:**
```text
   A
  ABA
 ABCBA
ABCDCBA
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
 A
ABA
```

---

## 2. Understanding the Logic

We use:
- One outer loop for rows.
- One space calculation.
- Two inner loops for increasing and decreasing alphabets.

### Step 1: Outer loop – Controls rows

```python
for i in range(1, n + 1):
```

For `n = 5`, `i` takes the values:

`1, 2, 3, 4, 5`

Each row adds one more alphabet on the left and right sides.

### Step 2: Calculate spaces

```python
space = n - i
print(" " * space, end="")
```

This centers the pyramid.

| Row (`i`) | Calculation | Spaces |
|---:|---|---:|
| 1 | 5 - 1 | 4 |
| 2 | 5 - 2 | 3 |
| 3 | 5 - 3 | 2 |
| 4 | 5 - 4 | 1 |
| 5 | 5 - 5 | 0 |

As the row number increases, spaces decrease.

### Step 3: Left side – Increasing alphabets

```python
for j in range(i):
    print(chr(65 + j), end="")
```

This prints alphabets from A up to the current row's alphabet.

| Row | `j` values | Output |
|---:|---|---|
| 1 | 0 | A |
| 2 | 0, 1 | AB |
| 3 | 0, 1, 2 | ABC |
| 4 | 0, 1, 2, 3 | ABCD |

Remember:
- `chr(65)` = A
- `chr(66)` = B
- `chr(67)` = C
- `chr(68)` = D

### Step 4: Right side – Decreasing alphabets

```python
for j in range(i - 2, -1, -1):
    print(chr(65 + j), end="")
```

This is the most important part of Pattern 17!

We need to print the alphabets backwards, but we must not repeat the middle alphabet.

**Why start from `i - 2`?**

The first loop has already printed the middle alphabet.

For example, when `i = 4`, the first loop prints:

`ABCD`

Now we only need:

`CBA`

So the second loop starts from `4 - 2 = 2`, which represents C.

Let's understand the range:

`range(i - 2, -1, -1)`

- Start: `i - 2`
- Stop: `-1` (not included)
- Step: `-1` (move backwards)

For `i = 4`:

`range(2, -1, -1)` gives:

`2, 1, 0`

Which produces:

`CBA`

### Step 5: Move to the next row

```python
print()
```

This moves to the next line after completing both alphabet loops.

## 3. Dry Run for n = 4

| Row | Spaces | Increasing side | Decreasing side | Final row |
|---:|---:|---|---|---|
| 1 | 3 | A | Nothing | A |
| 2 | 2 | AB | A | ABA |
| 3 | 1 | ABC | BA | ABCBA |
| 4 | 0 | ABCD | CBA | ABCDCBA |

### Final Output

```text
   A
  ABA
 ABCBA
ABCDCBA
```

---

## 4. Important Concept – Why `i - 2`?

Let's compare the two loops:

**First loop:**
```python
for j in range(i):
```
Prints: A, B, C, D

**Second loop:**
```python
for j in range(i - 2, -1, -1):
```
Prints: C, B, A

Together:

`ABCD + CBA = ABCDCBA`

We start from `i - 2` because D is already printed by the first loop.

---

## 5. Memory Trick

Remember: **Space → Up → Down**

1. Space: `n - i`
2. Left alphabet: `range(i)`
3. Right alphabet: `range(i - 2, -1, -1)`
4. Convert numbers into letters using `chr(65 + j)`.

**Key takeaway:** The first loop prints the middle alphabet. The second loop starts one position before it to avoid duplication.

---

## 6. Python Solution

```python
class Solution:
    def pattern17(self, n):
        for i in range(1, n + 1):
            space = n - i
            print(" " * space, end="")

            for j in range(i):
                print(chr(65 + j), end="")

            for j in range(i - 2, -1, -1):
                print(chr(65 + j), end="")

            print()
```
