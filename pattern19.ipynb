# Pattern 19 – Butterfly Pattern (Two-Sided Stars)

## 1. Problem Statement

Given an integer `n`, print a butterfly-like pattern using stars.

The pattern has two halves:

- Upper half: Stars decrease from `n` to `1`.
- Lower half: Stars increase from `1` to `n`.
- Middle spaces increase in the upper half and decrease in the lower half.

### Example 1

**Input:** `n = 4`

**Output:**
```text
********
***  ***
**    **
*      *
*      *
**    **
***  ***
********
```

### Example 2

**Input:** `n = 2`

**Output:**
```text
****
*  *
*  *
****
```

---

## 2. Understanding the Logic

We use two outer loops because the pattern has two different halves.

Each row prints:

`Left stars + Middle spaces + Right stars`

### Step 1: Upper half – Decreasing stars

```python
for i in range(1, n + 1):
    stars = n - i + 1
    spaces = 2 * (i - 1)
```

Let's understand both formulas.

**Stars formula:**

`stars = n - i + 1`

As `i` increases, the number of stars decreases.

**Spaces formula:**

`spaces = 2 * (i - 1)`

As `i` increases, the middle spaces increase by 2.

For `n = 4`:

| Row (`i`) | Stars calculation | Stars | Spaces calculation | Spaces |
|---:|---|---:|---|---:|
| 1 | 4 - 1 + 1 | 4 | 2 × (1 - 1) | 0 |
| 2 | 4 - 2 + 1 | 3 | 2 × (2 - 1) | 2 |
| 3 | 4 - 3 + 1 | 2 | 2 × (3 - 1) | 4 |
| 4 | 4 - 4 + 1 | 1 | 2 × (4 - 1) | 6 |

**Printing the row:**

```python
print("*" * stars + " " * spaces + "*" * stars)
```

This prints:
1. Left-side stars.
2. Middle spaces.
3. Right-side stars.

### Step 2: Lower half – Increasing stars

```python
for i in range(1, n + 1):
    stars = i
    spaces = 2 * (n - i)
```

Now the logic reverses.

**Stars formula:**

`stars = i`

As `i` increases, stars increase.

**Spaces formula:**

`spaces = 2 * (n - i)`

As `i` increases, middle spaces decrease.

For `n = 4`:

| Row (`i`) | Stars | Spaces calculation | Spaces |
|---:|---:|---|---:|
| 1 | 1 | 2 × (4 - 1) | 6 |
| 2 | 2 | 2 × (4 - 2) | 4 |
| 3 | 3 | 2 × (4 - 3) | 2 |
| 4 | 4 | 2 × (4 - 4) | 0 |

We use the same print statement:

```python
print("*" * stars + " " * spaces + "*" * stars)
```

---

## 4. Important Concept – Why Two Loops?

| Feature | Upper half | Lower half |
|---|---|---|
| Stars | Decrease | Increase |
| Spaces | Increase | Decrease |
| Stars formula | `n - i + 1` | `i` |
| Spaces formula | `2 * (i - 1)` | `2 * (n - i)` |

The second loop reverses the behavior of the first loop.

---

## 5. Memory Trick

Remember:

**Upper = Stars Down, Spaces Up**

**Lower = Stars Up, Spaces Down**

- Upper stars: `n - i + 1`
- Upper spaces: `2 * (i - 1)`
- Lower stars: `i`
- Lower spaces: `2 * (n - i)`

Both halves use the same print statement:

`"* " + spaces + "*"` (conceptually: left stars + spaces + right stars).

---

## 6. Python Solution

```python
class Solution:
    def pattern19(self, n):

        # Upper half
        for i in range(1, n + 1):
            stars = n - i + 1
            spaces = 2 * (i - 1)

            print("*" * stars + " " * spaces + "*" * stars)

        # Lower half
        for i in range(1, n + 1):
            stars = i
            spaces = 2 * (n - i)

            print("*" * stars + " " * spaces + "*" * stars)
```
